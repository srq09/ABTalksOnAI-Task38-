```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


documents = [
    "Artificial Intelligence enables machines to simulate human intelligence.",
    "Machine Learning is a subset of Artificial Intelligence.",
    "FAISS is a library used for vector similarity search.",
    "Vector databases store embeddings for semantic retrieval.",
    "Python is widely used in AI and Machine Learning.",
    "Large Language Models power modern AI applications.",
    "Retrieval Augmented Generation combines retrieval and generation.",
    "Embeddings convert text into numerical vector representations."
]

print("=" * 60)
print("VECTOR DATABASE SIMULATION")
print("=" * 60)

vectorizer = TfidfVectorizer()

document_vectors = vectorizer.fit_transform(documents)

while True:

    query = input("\nEnter your query (or type exit): ")

    if query.lower() == "exit":
        print("Program Closed.")
        break

    query_vector = vectorizer.transform([query])

    similarity_scores = cosine_similarity(
        query_vector,
        document_vectors
    )[0]

    ranked_results = sorted(
        enumerate(similarity_scores),
        key=lambda x: x[1],
        reverse=True
    )

    print("\nTop 3 Relevant Documents:\n")

    for rank, (index, score) in enumerate(
        ranked_results[:3],
        start=1
    ):
        print(f"Rank #{rank}")
        print(f"Document : {documents[index]}")
        print(f"Similarity Score : {score:.4f}")
        print("-" * 50)
```
